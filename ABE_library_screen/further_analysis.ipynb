```python
all_abe_samples_normalized_all_libs_edit_rate_file=f"{config.WORKDIR}/out/06_edit_rate/all-abe-samples-normalized_all-libs.edit_rate.csv"
all_abe_samples_normalized_all_libs_edit_rate=pd.read_csv(all_abe_samples_normalized_all_libs_edit_rate_file, index_col=0)
print(all_abe_samples_normalized_all_libs_edit_rate.shape)
all_abe_samples_normalized_all_libs_edit_rate.head(n=3)
```

```python
# Function to find spacer position allowing wildcard at target position
def find_spacer_position(spacer, target_pos, sequence):
    spacer_len = len(spacer)
    for i in range(len(sequence) - spacer_len + 1):
        match = True
        for j in range(spacer_len):
            # Skip comparison at target position (1-based to 0-based conversion)
            if j == target_pos - 1:
                continue
            if sequence[i + j] != spacer[j]:
                match = False
                break
        if match:
            return i
    return -1

# Reset the coding bystander column
all_abe_samples_normalized_all_libs_edit_rate['coding bystander'] = None
all_abe_samples_normalized_all_libs_edit_rate['strand coding bystander'] = None

# Iterate through each row to determine the coding bystander value
for idx, row in all_abe_samples_normalized_all_libs_edit_rate.iterrows():
    if pd.isna(row['lib_spacer']):
        continue
        
    if row['edit_position'] == row['target_position']:
        # If edit position equals target position, set to None
        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = None
    else:
        lib_spacer = row['lib_spacer']
        target_pos = row['target_position']
        edit_pos = row['edit_position']
        
        # Find position in MC4R_gene first
        pos_in_mc4r_gene = find_spacer_position(lib_spacer, target_pos, MC4R_gene)
        pos_in_mc4r_rc = find_spacer_position(lib_spacer, target_pos, MC4R_rc_gene)

        
        if pos_in_mc4r_gene != -1:
            # Found in MC4R_gene - check A->T mutation at edit position
            mc4r_pos = pos_in_mc4r_gene + edit_pos - 1 # (+edit_pos-1) is a group 
            
            if 0 <= mc4r_pos < len(MC4R_gene):
                # Create mutated sequence (original to T for A->G in ABE context)
                
                mutated_gene = list(MC4R_gene).copy()
                mutated_gene[mc4r_pos] = 'G'  # ABE changes A to G
                mutated_gene_str = ''.join(mutated_gene)
                
                
                # Translate both sequences
                original_protein = Seq(MC4R_gene).translate()
                mutated_protein = Seq(mutated_gene_str).translate()
                
                # Check if amino acid changes
                all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = str(original_protein) != str(mutated_protein)
                all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="+"
                
                
                # print("pos_in_mc4r_gene, pos_in_mc4r_rc",pos_in_mc4r_gene, pos_in_mc4r_rc)
                # break
                
        elif pos_in_mc4r_rc != -1:
            # Found in MC4R_rc_gene - calculate corresponding position in MC4R_gene
            mc4r_pos = len(MC4R_gene)- 1 - (pos_in_mc4r_rc + edit_pos - 1) 
            
            
            
            if 0 <= mc4r_pos < len(MC4R_gene):
                # Create mutated sequence (C mutation on reverse complement)
                mutated_gene = list(MC4R_gene).copy()
                mutated_gene[mc4r_pos] = 'G'
                mutated_gene_str = ''.join(mutated_gene)
                
                
                # Translate both sequences
                original_protein = Seq(MC4R_gene).translate()
                mutated_protein = Seq(mutated_gene_str).translate()
                
                # Check if amino acid changes
                all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = str(original_protein) != str(mutated_protein)
                all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="-"
                
                
                # print("pos_in_mc4r_gene, pos_in_mc4r_rc",pos_in_mc4r_gene, pos_in_mc4r_rc)
                # break

        else: 
            # Not found in either sequence - this shouldn't happen for valid data
            pos_in_mc4r_full_gene = find_spacer_position(lib_spacer, target_pos, full_MC4R_gene)
            pos_in_mc4r_full_rc = find_spacer_position(lib_spacer, target_pos, full_MC4R_rc_gene)

            if (pos_in_mc4r_full_gene == -1) and (pos_in_mc4r_full_rc == -1):
                raise ValueError(f"idx {idx} Spacer '{lib_spacer}' with target_position {target_pos} not found in either MC4R_gene or MC4R_rc_gene for row {idx}")
            else:
                # Check if spacer is in junction/flanking region or completely outside CDS
                if pos_in_mc4r_full_gene != -1:
                    # Calculate MC4R_gene boundaries within full_MC4R_gene
                    mc4r_start = full_MC4R_gene.find(MC4R_gene)
                    mc4r_end = mc4r_start + len(MC4R_gene)
                    
                    spacer_start = pos_in_mc4r_full_gene
                    spacer_end = pos_in_mc4r_full_gene + len(lib_spacer)
                    
                    # Check if spacer overlaps with MC4R_gene boundaries
                    if (spacer_start < mc4r_end and spacer_end > mc4r_start):
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = "In_Junction"
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="+"
                        # print("pos_in_mc4r_gene, pos_in_mc4r_rc",pos_in_mc4r_gene, pos_in_mc4r_rc)
                        # print("mc4r_pos", mc4r_pos)
                        # break
                    
                    else:
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = "outside_CDS"
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="+"
                
                elif pos_in_mc4r_full_rc != -1:
                    # Calculate MC4R_rc_gene boundaries within full_MC4R_rc_gene
                    mc4r_rc_start = full_MC4R_rc_gene.find(MC4R_rc_gene)
                    mc4r_rc_end = mc4r_rc_start + len(MC4R_rc_gene)
                    
                    spacer_start = pos_in_mc4r_full_rc
                    spacer_end = pos_in_mc4r_full_rc + len(lib_spacer)
                    
                    # Check if spacer overlaps with MC4R_rc_gene boundaries
                    if (spacer_start < mc4r_rc_end and spacer_end > mc4r_rc_start):
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = "In_Junction"
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="-"
                    else:
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'coding bystander'] = "outside_CDS"
                        all_abe_samples_normalized_all_libs_edit_rate.at[idx, 'strand coding bystander']="-"
      

    

print("Updated 'coding bystander' column")
#2 mins
```

```python
# Create long format dataframe for all_abe_samples_normalized_all_libs_edit_rate
result_rows = []

# Group by editor, site_id, sublibrary, and replicate
grouped = all_abe_samples_normalized_all_libs_edit_rate.groupby(['editor', 'site_id', 'sublibrary', 'replicate'])

for (editor, site_id, sublibrary, replicate), group in grouped:
    # Get common values for the group
    pam = group['PAM'].iloc[0]
    lib_spacer = group['lib_spacer'].iloc[0]
    target_position = group['target_position'].iloc[0]
    
    # Initialize row data
    row_data = {
        'editor': editor,
        'site_id': site_id, 
        'sublibrary': sublibrary,
        'PAM': pam,
        'lib_spacer': lib_spacer,
        'replicate': replicate,
        'On-target base position': None,
        'On-target base position editing rate': None
    }
    
    # Find on-target position
    on_target = group[group['edit_position'] == target_position]
    if not on_target.empty:
        row_data['On-target base position'] = target_position
        row_data['On-target base position editing rate'] = on_target['edit_rate'].iloc[0]
    
    # Find bystander positions
    bystanders = group[group['edit_position'] != target_position].sort_values('edit_position')
    
    for i, (_, bystander_row) in enumerate(bystanders.iterrows(), 1):
        row_data[f'Bystander position {i}'] = bystander_row['edit_position']
        row_data[f'Bystander position {i} editing rate'] = bystander_row['edit_rate']
        row_data[f'Bystander position {i} coding bystander'] = bystander_row['coding bystander']
    
    result_rows.append(row_data)

# Convert to DataFrame
all_abe_samples_normalized_all_libs_edit_rate_wide_format_df = pd.DataFrame(result_rows)
all_abe_samples_normalized_all_libs_edit_rate_wide_format_df.head(n=3)

#20 seconds 
```

```python
# Merge the precise editing rates with the wide format dataframe
merged_df = all_abe_samples_normalized_all_libs_edit_rate_wide_format_df.merge(
    all_abe_samples_normalized_all_libs_edit_rate_precise[['editor', 'site_id', 'sublibrary', 'replicate', 'edit_position', 'edit_rate']],
    left_on=['editor', 'site_id', 'sublibrary', 'replicate', 'On-target base position'],
    right_on=['editor', 'site_id', 'sublibrary', 'replicate', 'edit_position'],
    how='left'
)

# Rename the edit_rate column to the desired name and drop the redundant edit_position column
merged_df = merged_df.rename(columns={'edit_rate': 'Only precise editing (only on-target base)'})
merged_df = merged_df.drop(columns=['edit_position'])

# Update the original dataframe
all_abe_samples_normalized_all_libs_edit_rate_wide_format_df = merged_df
all_abe_samples_normalized_all_libs_edit_rate_wide_format_df.head(n=3)
```
